```python
#adding an element to set
student={"amit","ganesh","ram","sanjeev"}
student.add("priya")
print(student)
```

```python
#adding multiple element to set
student_name={"amit","priya"}
student_name.update(["parul","kiran","sneha","neha"])
print(student_name)
```

```python
#removing item from set
student={"amit","ganesh","ram","sanjeev"}
student.remove("amit")
print(student)
#remove only the element preseent in the set and if the element is not present in tbe set it throws an erroe
```

```python
#discarding item from set
student={"amit","ganesh","ram","sanjeev"}
student.discard("riya")
print(student)
#remove  the element preseent that may not be present in the set
```

```python
#pop function
nums={10,20,30,40}
x=nums.pop()
print(x)
print(nums)
#removes last element
```

```python
#clear function
#only the values get deleted , you can access the set
nums={10,20,30,40}
nums.clear()
print(nums)
```

```python
#delete a set
nums={10,20,30,40}
del nums
print(nums)
#delete the complete set
```
